```python
def inspect_meta_scores(df, n=5):
    """
    상위 n개의 세션에 대해 4대 메타 축(Volume, Size_Payload, Topology, Control_State)의 
    실제 점수가 어떻게 집계되었는지 확인합니다.
    """
    print(f"\n================ [상위 5개 세션 메타 축 점수 검증] ================")
    
    # 4대 메타 축 스코어 컬럼 (데이터셋에 실제 존재하는지 확인)
    meta_cols = ['Volume_Score', 'Size_Payload_Score', 'Topology_Score', 'Control_State_Score']
    
    # 만약 데이터에 메타 축 스코어가 아직 없다면 매핑 정의를 이용해 계산하는 로직 보완 가능
    # 여기서는 이미 마스터 데이터셋에 해당 스코어 컬럼들이 있다고 가정하고 출력합니다.
    available_cols = [col for col in meta_cols if col in df.columns]
    
    if not available_cols:
        print("[WARNING] 데이터셋에 4대 메타 축 스코어 컬럼이 존재하지 않습니다.")
        return

    # 상위 n개 추출
    sample_df = df.head(n).copy()
    
    # 확인용 출력 데이터프레임 구성
    inspection_columns = ['Year', 'Attack_Group', 'Label'] + available_cols if 'Year' in df.columns else available_cols
    existing_inspection_cols = [c for c in inspection_columns if c in sample_df.columns]
    
    print(sample_df[existing_inspection_cols].to_string(index=True))
    print("====================================================================\n")

# 실행 테스트
if 'df_master' in locals() and not df_master.empty:
    inspect_meta_scores(df_master, n=5)
else:
    print("[ERROR] df_master 데이터가 로드되지 않았습니다.")
```
